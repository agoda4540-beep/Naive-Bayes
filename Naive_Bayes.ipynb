```python
import pandas as pd
import numpy as np

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer


from sklearn.naive_bayes import BernoulliNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score
```

```python
train_data = fetch_20newsgroups(subset='train')
test_data = fetch_20newsgroups(subset='test')
```

```python
x_train = train_data.data
y_train = train_data.target

x_test = test_data.data
y_test = test_data.target
```

Start BernoulliNB

```python
vec_bern = CountVectorizer(binary = True, max_features= 10000)
x_train_vec = vec_bern.fit_transform(x_train)
x_test_vec = vec_bern.transform(x_test)
```

```python
model_bern = BernoulliNB()
model_bern.fit(x_train_vec, y_train)
```

```python
pred_bern = model_bern.predict(x_test_vec)
```

```python
print(x_train_vec.shape)
print(x_test_vec.shape)
```

```python
acc_bern = accuracy_score(y_test, pred_bern)
print(acc_bern)
```

End BernoulliNB

********************************************************************************

Start MultinomialNB

```python
vec_multi = CountVectorizer(max_features= 10000)
x_train_multi = vec_multi.fit_transform(x_train)
x_test_multi = vec_multi.transform(x_test)
```

```python
model_multi = MultinomialNB()
model_multi.fit(x_train_multi, y_train)
```

```python
pred_multi = model_multi.predict(x_test_multi)
```

```python
print(x_train_multi.shape)
print(x_test_multi.shape)
```

```python
acc_multi = accuracy_score(y_test, pred_multi)
print(acc_multi)
```

End MultinomialNB

********************************************************************************

Start Gaussian

```python
vec_gauss = TfidfVectorizer(max_features= 2000)

x_train_gauss = vec_gauss.fit_transform(x_train).toarray()
x_test_gauss = vec_gauss.transform(x_test).toarray()
```

```python
model_gauss = GaussianNB()
model_gauss.fit(x_train_gauss, y_train)
```

```python
pred_gauss = model_gauss.predict(x_test_gauss)
```

```python
print(x_train_gauss.shape)
print(x_test_gauss.shape)
```

```python
acc_gauss = accuracy_score(y_test, pred_gauss)
print(acc_gauss)
```

End Gaussian

```python
data = {
    "Model" : ["BernoulliNB", "MultinomialNB", "GaussianNB"],
    'Data Representation': ['Binary word presence', 'Word counts', 'Numerical (TF-IDF)'],
    "Accuracy" : [acc_bern, acc_multi, acc_gauss]
}

df = pd.DataFrame(data)
df
```

a. Why is each representation suitable for its model ?

BernoulliNB: It works with binary data (0/1), making it suitable for checking if a word is present in the text or not.

MultinomialNB: It uses word counts, which helps capture word frequency to better classify the text categories.

GaussianNB: It requires continuous numerical values (like TF-IDF scores) that fit a normal (Gaussian) distribution.

********************************************************************************

b. Which model achieved the highest accuracy ?

MultinomialNB achieved the highest accuracy (77.0%), because word count frequency provides the strongest signals for classifying text into news categories.

********************************************************************************

c. Does the best accuracy always mean the model is theoretically the most
suitable ?

No, higher accuracy on a specific test does not always mean theoretical superiority. Accuracy is an empirical metric that depends on data size and parameters, while theoretical suitability depends on whether the model's mathematical assumptions align with the actual data structure.
